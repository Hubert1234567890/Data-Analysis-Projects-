production summary

```python
production['Date'] = pd.to_datetime(
    production['Date'],
    dayfirst=True
)
production['Month'] = production['Date'].dt.month_name()
production[['Date', 'Month']].head()
```

```python
production_summary = production.groupby(
    ['Month', 'Category']
).agg(
    Total_Produced=('Total Products', 'sum'),
    Total_Defective=('Defective Products', 'sum')
).reset_index()
production_summary
```

```python
## calculating the good products
production_summary['Good_Products'] = (
    production_summary['Total_Produced']
    - production_summary['Total_Defective']
)
## calculating the defect rate
production_summary['Defect_Rate_%'] = (
    production_summary['Total_Defective']
    / production_summary['Total_Produced']
) * 100
production_summary
```

```python
retail['Date'] = pd.to_datetime(
    retail['Date'],
    dayfirst=True
)
retail['Month'] = retail['Date'].dt.month_name()
retail['Category'] = retail['Category'].str.strip().str.title()
retail['Return Status'].value_counts()
```

```python
retail['Net Sales Amount'] = retail['Total Amount'].where(
    retail['Return Status'] == 'Kept',
    0
)
retail['Net Quantity Sold'] = retail['Quantity'].where(
    retail['Return Status'] == 'Kept',
    0
)
retail['Returned Quantity'] = retail['Quantity'].where(
    retail['Return Status'] == 'Returned',
    0
)
retail.head()
```

```python
retail_summary = retail.groupby(
    ['Month', 'Category']
).agg(
    Units_Sold=('Net Quantity Sold', 'sum'),
    Returned_Units=('Returned Quantity', 'sum'),
    Gross_Revenue=('Total Amount', 'sum'),
    Net_Revenue=('Net Sales Amount', 'sum')
).reset_index()
retail_summary
```

stall sales

```python
stall_sales['Date'] = pd.to_datetime(
    stall_sales['Date'],
    dayfirst=True
)

stall_sales['Month'] = stall_sales['Date'].dt.month_name()
stall_sales['Category'] = (
    stall_sales['Category']
    .str.strip()
    .str.title()
)
stall_sales['Net Revenue'] = (
    stall_sales['Total Amount'] *
    (1 - stall_sales['Discount (%)'] / 100)
)
stall_sales.head()
```

```python
stall_summary = stall_sales.groupby(
    ['Month', 'Category']
).agg(
    Units_Sold=('Quantity', 'sum'),
    Gross_Revenue=('Total Amount', 'sum'),
    Net_Revenue=('Net Revenue', 'sum')
).reset_index()
stall_summary
stall_sales[['Total Amount', 'Discount (%)', 'Quantity']].head(10)
```

```python
production['Variable Cost per Unit'] = (
    production['Cost of Production (per unit)']
    + production['Wages (per unit)']
)
production.head()
```

```python
production_cost = production.groupby(
    ['Month', 'Category']
).agg(
    Avg_Production_Cost=(
        'Cost of Production (per unit)', 'mean'
    ),
    Avg_Wages=(
        'Wages (per unit)', 'mean'
    )
).reset_index()
production_cost
```

combined sales

```python
sales_summary = pd.concat(
    [retail_summary, stall_summary],
    ignore_index=True
)
sales_summary ## returned units have nan values because the the stall sales have no retrun
```

```python
total_sales = sales_summary.groupby(
    ['Month', 'Category']
).agg(
    Units_Sold=('Units_Sold', 'sum'),
    Net_Revenue=('Net_Revenue', 'sum')
).reset_index()
total_sales
```

```python
final = total_sales.merge(
    production_cost,
    on=['Month', 'Category'],
    how='left'
)
final
```

operational summary

```python
operations['Date'] = pd.to_datetime(
    operations['Date'],
    dayfirst=True
)
operations['Month'] = operations['Date'].dt.month_name()
monthly_operations = operations.groupby(
    'Month'
).agg(
    Total_Operating_Expense=('Amount', 'sum')
).reset_index()
monthly_operations
```

```python
total_sales
```

```python
# Merge sales summary with production cost
financial_data = total_sales.merge(
    production_cost,
    on=['Month', 'Category'],
    how='left'
)

# Calculate COGS
financial_data['COGS'] = (
    financial_data['Units_Sold']
    * financial_data['Avg_Production_Cost']
)

# Calculate Gross Profit
financial_data['Gross_Profit'] = (
    financial_data['Net_Revenue']
    - financial_data['COGS']
)

# Merge monthly operating expenses
financial_data = financial_data.merge(
    monthly_operations,
    on='Month',
    how='left'
)

# Monthly financial summary
monthly_financial = financial_data.groupby(
    'Month'
).agg(
    Total_Revenue=('Net_Revenue', 'sum'),
    COGS=('COGS', 'sum'),
    Gross_Profit=('Gross_Profit', 'sum'),
    Total_Operating_Expense=('Total_Operating_Expense', 'first')
).reset_index()

# Calculate Net Profit
monthly_financial['Net_Profit'] = (
    monthly_financial['Gross_Profit']
    - monthly_financial['Total_Operating_Expense']
)

# Display final result
monthly_financial
```
